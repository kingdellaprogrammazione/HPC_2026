```python
import matplotlib.pyplot as plt
import matplotlib as mpl
import numpy as np
import seaborn as sns
```

```python
x = np.arange(10)


series = np.array([
    [10,12,15,14,18,21,19,23,25,28],
    [5,7,6,10,12,15,14,18,20,22],
    [20,18,22,25,27,26,30,32,35,37],
    [15,17,16,20,22,21,25,27,26,30],
    [30,28,31,29,35,33,36,34,38,40],
])

average = series.mean(axis=0)
```

```python
def beeswarm_offsets(y, spacing=0.08):
    """Simple horizontal offsets to avoid overlapping points."""
    order = np.argsort(y)
    offsets = np.zeros(len(y))
    direction = 1

    for i, idx in enumerate(order):
        offsets[idx] = direction * spacing * ((i + 1) // 2)
        direction *= -1

    return offsets
```

```python
fig, ax = plt.subplots(figsize=(3.39, 2.2), constrained_layout=True)

ax.set_axisbelow(True)

plt.grid(
    color="0.85",
    linestyle="-",
    linewidth=1,
    alpha=1.0
)

for i in range(len(x)):
    values = series[:, i]
    offsets = beeswarm_offsets(values)

    plt.scatter(
        np.full(len(values), x[i]) + offsets,
        values,
        s=10,
        alpha=0.8
    )

# add average line
avg = series.mean(axis=0)
plt.plot(x, avg, color="black", marker="o",     markersize=4,  linewidth=2, label="Mean")

plt.xlabel("Grid side $M$")
plt.ylabel("Time (s)")
#plt.title("MPI Timings")
leg = ax.legend(
    loc="upper left",
    fontsize=7,
    frameon=True,
    fancybox=False,
    framealpha=1.0,
    facecolor="1.0",
    edgecolor="0.85",
    borderpad=0.5,
)

leg.get_frame().set_linewidth(1)

# plt.show()
plt.savefig("./images/mpi_scaling.pgf")
plt.savefig("./images/mpi_scaling.pdf")
```
